# Zomato data analysis project

# step 1 - Importing libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# step 2 - Creat the data frame

```python
df = pd.read_csv ("Zomato data  (1).csv")
print (df)
```

```python
df.head()
```

# Convert the data type of column -rate

```python
def handleRate(value):
    value = str(value).split('/')
    value= value[0];
    return float (value)
    
df['rate'] =df['rate'].apply(handleRate)
print(df.head())
```

```python
df.head()
```

# Type of returant

```python
df.head()
```

```python
sns.countplot(x=df['listed_in(type)'])
plt.xlabel("type of resturant")
```

# conclusion - majority of the returant falls in dining category

```python
df.head ()
```

```python
grouped_data = df.groupby ('listed_in(type)')['votes'].sum()
result = pd.DataFrame({'votes'  : grouped_data})
plt.plot(result, c="green", marker="o")
plt.xlabel('Type of Restaurant', color='red', size=20)
plt.ylabel('Votes', color='red', size=20)
```

# conclusion - dining returant has recieved maximum votes

```python
plt.hist(df['rate'],bins =4)
plt.title("ratings distribution")
plt.show()
```

# conclusion - the majority returants recieved ratings from 3.5 to 4

# Average order spending by couples

```python
df.head()
```

```python
couple_data=df['approx_cost(for two people)']
sns.countplot(x=couple_data)
```

# conclusion - the majority of couples prefer resturant with an approximate cost of 300 rupees

# wich mode recieves maximum rating

```python
df.head ()
```

```python
plt.figure (figsize =(6,6))
sns.boxplot(x = 'online_order', y = 'rate', data = df)
```

# conclusion - offline order recieved lower rating  in comaprison to online order

# pivot tabel

```python
df.head()
```

```python
pivot_table = df.pivot_table(index='listed_in(type)', columns='online_order', aggfunc='size', fill_value=0)
sns.heatmap(pivot_table, annot=True, cmap='YlGnBu', fmt='d')
plt.title("Heatmap")
plt.xlabel("Online Order")
plt.ylabel("Listed In (Type)")
plt.show()
```

# CONCLUSION: Dining restaurants primarily accept offline orders, whereas cafes primarily receive online orders. This suggests that clients prefer orders in person at restaurants, but prefer online ordering at cafes.
